# Exceptions and Debugging

Python raises an exception whenever it tries to execute an incorrect/invalid code.

We can raise our own exceptions using the **raise** statement.

An exception consists of : 
- raise keyword
- a call to the Exception() function
- A string with a helpful error message passed to the Exception() function

````python
raise Exception("This is an error, check the validity of the dates")
````

````python
def boxPrint(symbol, width, height):
    if len(symbol) != 1:
        raise Exception("Symbol must be a single character string.")
    if not isinstance(symbol, str):
        raise Exception("Symbol must be a string")
    if width <= 2:
        raise Exception("Width must be greater than 2.")
    if height <= 2:
        raise Exception("Height must be greater than 2.")
    print(symbol * width)

    for i in range(height -2):
        print(symbol + (' '* (width -2)) + symbol)
    print(symbol * width)
    
````

````python
## We use the try and except block to recover from the errors we might have anticipated

for sym, w, h in (('zz',3,3),('*',4,4),('0',20,5),('x',4,3)):
    try:
        boxPrint(sym, w, h)
    except Exception as err:
        print('An error has been encountered: ' +str(err))
````

We also have a **try ... except ... else ... finally** block of statements for error handling. For example: 

```
try:
    getfile = open("filename.txt","r")
    getfile.write("My file for exception handling")
except IOError:
    print("Unable to open the data in the file")
else:
    print("The file was written successfuly")
finally:
    getfile.close()
    print("File has now been closed")
```    

The finally block is executed no matter the outcome of the code.

The try and except block helps us handle errors more gracefully instead of letting the whole program crash.

````python
def spam():
    bacon()

def bacon():
    raise Exception('This is an error message')
````

````python
spam()
````

This trove of information is known as the **traceback**. And the sequence of function calls that led to the error is called **call stack**.

Even when we have the try and except blocks with raise embedded, we can use the traceback module to obtain the traceback as a string.

````python
try:
    result = 9/0
except ZeroDivisionError:
    print("Cannot divide real number by zero")
````

````python
import traceback
try:
    result = 9/0
except ZeroDivisionError:
    print(traceback.format_exc())
    #print("Cannot divide real numbers by zero")
````
