```python
### The mean yield for each crop type
```

```python
crops = data.crop_type_main.unique()
yield_means = []

for crop in crops:
    mean_ = float(data.loc[data.crop_type_main == crop].Yield_main_in_Kg.mean())
    yield_means.append(mean_)

crop_yield_means = dict(zip(crops, yield_means))
print(crop_yield_means)
```

```python
crops = list(crop_yield_means.keys())
yields = list(crop_yield_means.values())
```

```python
mean_yields_df = pd.DataFrame({"Crops":crops, "Mean Yield":yields})
```

```python
plt.figure(figsize=(10,10))
plt.pie(mean_yields_df["Mean Yield"], labels= mean_yields_df.Crops, autopct="%0.2f%%")
plt.show()
```
