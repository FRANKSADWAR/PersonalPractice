The self constructor
And the init method used to initialize a class.

These are the two conerstones of object oriented programming in Python.

The init method is used by classes in Python to initialize the class by setting initial values for instance attributes.

**self** refers to an instance of the class object.

```python
import matplotlib.pyplot as plt
import math

%matplotlib inline
```

```python
class Circle(object):
    """
    Represents a circle objects
    Attributes: 
        color: str
        radius: float
    """
    def __init__(self, color, units='cm', radius = 1):
        self.color = color
        self.radius = radius
        self.units = units

    def get_area(self):
        """
        Gets the area of the circle
        """
        area = round((math.pi * self.radius * self.radius),3)
        print (f"The area of the circle is {area} {self.units} squared ")
        return area
        

    def get_perimeter(self):
        """
        Gets the circumference of the circle
        """
        perim = math.pi * 2 * self.radius
        print(f"The perimeter of the circle is {preim} {self.units}")
        return preim
        

    def add_radius(self, r):
        self.radius = self.radius + r
        return self.radius

    def draw_circle(self):
        plt.gca().add_patch(plt.Circle((0,0), radius = self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()

        
```

```python
class Person:
    def __init__(self, name, job, pay=1000, currency='USD'): ### The constructor method for initializing a class
        self.name = name
        self.job = [job]
        self.pay = pay
        self.currency = currency
    
    def __repr__(self): ## Operator overloading to intercept print operations in the __repr__ method
        return f'[Person : {self.name}, {self.job}]'
        
    def __len__(self):  ## A special type of operator overloading
        return f'Jobs: {len(self.job)}'
        
    def last_name(self):
        return self.name.split()[-1]
   
    def give_raise(self, percent):
        self.new_pay = int(self.pay * (1 + percent))
    
```

```python
class Manager(Person):
    def give_raise(self, percent, bonus = 0.1):
        Person.give_raise(self, percent + bonus)
        print(f"Previous salary: {self.pay} New salary: {self.new_pay}")
```

In the class Manager above, we have inherited from the Person class, and extended the give_raise method, to include
a bonus for manager's pay computation.

In inhreritance with Python, the lowest version of the name wins.

```python
from __future__ import print_function
from datetime import datetime

class Employee:
    """
    Represents an employee with a name and salary.

    Attributes:
        name (str):  The employee's name
        
        salary (float: The salary of the employee.
        
    Methods:
        give_raise (percent): The percentage with which to increase the employee's salary.

        work() : simulates what the employee does by printing it out

        __repr__(): Returns a deveoper friendly string representation of the Employee object, 
                    showing the employee name and salary.
    """
    def __init__(self, name, salary = 0):
        self.name = name
        self.salary = salary

    def give_raise(self, percent):
        """
        Increases the employee's sakary by the percentage parsed.

        Args:
            percent (float): the percentage with shich to increase the employee's salary
        Returns:
            Returns the new employee's salary after applying the increase
        """
        
        self.salary = self.salary + (self.salary * percent)

    def work(self):
        print(self.name, "Does stuff")

    def __repr__(self):
        return f"<Employee: {self.name} salary: {self.salary}>"
```

```python
class Chef(Employee): ## Chef class inherits Employee
    def __init__(self, name):
        Employee.__init__(self, name, 90000)
        
    def work(self):
        Employee.work(self)
        print(self.name, "Make food")
        
    def __repr__(self):
        return f"<Chef: {self.name} and salary: {self.salary}>"
```

```python
class Driver(Employee):
    def __init__(self, name):
        super().__init__(name, 90000)

    def actions(self, date):
        date = date.strftime("%Y-%m-%d %H:%M:S")
        routes = ['Utawala', 'Fedha','Malaa']
        for i in routes:
            print(f"Taking route {i} on {date}")

        return {"date":date, "routes":routes}
```

```python
class TextAnalyzer(object):

    def __init__(self, text):
        formattedText = text.replace(",","").replace(".","").replace("!","").replace("?","")
        formattedText = formattedText.lower()
        self.fmtText = formattedText

    def freq_all(self):
        words_list = self.fmtText.split()

        freq_map = {}

        for word in set(words_list):
            freq_map[word] = words_list.count(word)

        return freq_map
        
    def freq_word(self, word):
        freq_map = self.freq_all()

        if word in freq_map:
            return freq_map[word]
        else:
            return 0
```
